```python
#accepts a sequence of comma-seprated numbers from user and generate a tuple with those numbers and find average of all items in the tuple
numbers = input("Enter comma-separated numbers: ")
num_tuple = tuple(map(int, numbers.split(",")))
average = sum(num_tuple) / len(num_tuple)
print("Tuple:", num_tuple)
print("Average:", average)
```
