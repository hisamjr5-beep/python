```python
#write a python program to get the 3rd element and 3rd element from last of a tuple
numbers=input("enter tuple elements separated by comma:")
tuple1=tuple(int(x)for x in numbers.split(","))
print("3rd element:",tuple1[2])
print("3rd element from list:",tuple1[-3])
```

```python
#accepts a sequence of comma-seprated numbers from user and generate a tuple with those numbers and find average of all items in the tuple
numbers = input("Enter comma-separated numbers: ")
num_tuple = tuple(int(x)for x in numbers.split(","))
average = sum(num_tuple)/len(num_tuple)
print("Tuple:",num_tuple)
print("Average:",average)
```
