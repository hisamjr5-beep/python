```python
#write a python program which accepts a sequence of comma-separated numbers from user and generate a list with those numbers and find sum of all items in thath list
numbers=input("eneter comma-separated numbers: ").split(",")
numbers =[int(x)for x in numbers]
print("list:",numbers)
print("sum=",sum(numbers))
```
